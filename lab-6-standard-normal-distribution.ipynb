# HISTOGRAM

```python
plt.hist(df['Current'], bins=5, edgecolor='black',color = 'skyblue', alpha=0.8)

plt.title('Histogram of Current Values', fontsize=16, fontfamily='Montserrat')
plt.xlabel('Current (mA)')
plt.ylabel('Frequency')
plt.show()
```

# NORMAL DISTRIBUTION

```python
mean = df['Current'].mean()
std = df['Current'].std(ddof=1)

x_min = df['Current'].min()
x_max = df['Current'].max()

x = np.linspace(x_min, x_max, 100)
p = stats.norm.pdf(x, mean, std)    

plt.plot(x, p, 'k', linewidth=2,color='orange')

plt.title('Normal Distribution Curve of Current Values', fontsize=16, fontfamily='Montserrat')
plt.xlabel('Current (mA)')
plt.ylabel('Probability Density')
```

```python
plt.hist(df['Current'], bins=5, edgecolor='black',color = 'skyblue', alpha=0.8, density=True)

plt.title('Histogram of Current Values', fontsize=16, fontfamily='Montserrat')

    

plt.plot(x, p, 'k', linewidth=2,color='orange')

plt.title('Normal Distribution Curve of Current Values ', fontsize=16, fontfamily='Montserrat')
plt.xlabel('Current (mA)')
plt.ylabel('Probability Density')
```

# Standard Normal Distribution

```python
mean = df['Current'].mean()
std = df['Current'].std(ddof=1)

x_min = df['Z-score'].min()
x_max = df['Z-score'].max()

x = np.linspace(x_min, x_max, 100)
p = stats.norm.pdf(x, df['Z-score'].mean(), df['Z-score'].std(ddof=1))   

plt.plot(x, p, 'k', linewidth=2,color='orange')

plt.title('Standard Normal Distribution Curve of Current Values N(0,1)', fontsize=16, fontfamily='Montserrat')
plt.xlabel('Z-score')
plt.ylabel('Probability Density')
```
