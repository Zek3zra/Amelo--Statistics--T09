```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv(r'dataset\current-power.csv',delimiter= ",")

df
```

```python
df.describe()
```

```python
df.cov()
```

```python
x = df['Current']
y = df['Power']

plt.scatter(x, y)
plt.title('Device Current Vs Power Response')
plt.xlabel('Current (A)')
plt.ylabel('Power (W)')
plt.grid()
```

```python
df.corr()
```

```python
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
line = slope * x + intercept

plt.figure(figsize=(7, 4))
plt.scatter(x,y,
            color='blue',
            alpha=0.6,
            label = "Data Points")  
plt.plot(x, line,
            color='red',
            label=f'Best Fit Line(r = {r_value:.3f})')
plt.grid()
plt.xlabel('Current (A)')
plt.ylabel('Power (W)')
plt.title('Linear Regression: Device Current Vs Power Response')
plt.legend()
```

## Air Quality

```python
air = pd.read_csv(r'dataset\air-quality.csv',delimiter= ",")

air
```

## NOx vs Humidity

```python
x = air['NOx']
y = air['Relative_Humidity']

slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
line = slope * x + intercept

plt.figure(figsize=(7, 4))
plt.scatter(x,y,
            color='blue',
            alpha=0.2,
            label = "Data Points")  
plt.plot(x, line,
            color='red',
            label=f'Best Fit Line(r = {r_value:.3f})')
plt.grid()
plt.xlabel('NOx (ug/cu.m)')
plt.ylabel('Relative Humidity (%)')
plt.title('Linear Regression: NOx vs Relative Humidity')
plt.legend()
```

## NO2 vs Humidity

```python
x = air['NO2']
y = air['Relative_Humidity']

slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
line = slope * x + intercept

plt.figure(figsize=(7, 4))
plt.scatter(x,y,
            color='blue',
            alpha=0.2,
            label = "Data Points")  
plt.plot(x, line,
            color='red',
            label=f'Best Fit Line(r = {r_value:.3f})')
plt.grid()
plt.xlabel('NO2 (ug/cu.m)')
plt.ylabel('Relative Humidity (%)')
plt.title('Linear Regression: NO2 vs Relative Humidity')
plt.legend()
```
