```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv(r'dataset\transformer-voltage.csv',delimiter= ",", index_col=0)

df
```

```python
df.info()
```

```python
df.describe()
```

```python
alpha = 0.05
z_critical = stats.norm.ppf(1-alpha/2)
z_critical
```

```python
pop_std = 5

ci_lower , ci_upper = stats.norm.interval (
    confidence = 0.95,
    loc = df ['Voltage'].mean(),
    scale = pop_std/np.sqrt(df['Voltage'].count())
)

print(f'CI: {ci_lower:.2f} and {ci_upper:.2f}  ')
```

```python



mean = df['Voltage'].mean()
std = df['Voltage'].std(ddof=1)

x_min = df['Voltage'].min()
x_max = df['Voltage'].max()

x = np.linspace(x_min, x_max, 100)
p = stats.norm.pdf(x, mean, std)    

plt.plot(x, p, 'k', linewidth=2,color='orange')


plt.hist(df['Voltage'], bins=5, density=True, alpha=0.6, color='b')


plt.title('Voltage distribution', fontsize=16, fontfamily='Montserrat')
plt.xlabel('Voltage(V)')
plt.ylabel('Probability Density')

plt.axvline(x = ci_lower,
            color = 'black',
            linestyle = '--' )

plt.axvline (x = ci_upper,
             color = 'black',
             linestyle = '--' )

plt.show()
```
