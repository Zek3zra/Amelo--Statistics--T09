# Measures of central tendency

```python
#Jaguar Vs Panthers
resistance = np.array([
    [10.6, 9.1, 9.3, 9.8, 10.5, 10.4, 9.5,11,10.4, 3, 9.8],
    [10.1, 11, 9.1, 20, 9.2, 10.8, 9.9, 9.2 , 9.1 ,9.1, np.nan]
])
np.shape(resistance)
```

```python
res_mean = np.nanmean(resistance, axis=1)
print("Jaguar Mean: " + str((res_mean[0]).round(2)))
print("Panthers Mean: " + str((res_mean[1]).round(2)))
```

```python
res_med = np.nanmedian(resistance, axis=1)
print("Jaguar Median: " + str((res_med[0]).round(2)))
print("Panthers Median: " + str((res_med[1]).round(2)))
```

```python
pan_res = resistance[1]
print("Panthers Mode:") 
for i in multimode(pan_res):
    print(f'{i} ohms')
```

```python
jag_res = resistance[0]
print("Jaguar Mode:") 
for i in multimode(jag_res):
    print(f'{i} ohms')
```

# Measures of Variability

```python
#Jaguar Vs Panthers
resistance = np.array([
    [10.6, 9.1, 9.3, 9.8, 10.5, 10.4, 9.5,11,10.4, 3, 9.8],
    [10.1, 11, 9.1, 20, 9.2, 10.8, 9.9, 9.2 , 9.1 ,9.1, np.nan]
])
np.shape(resistance)
```

```python
res_max = np.nanmax(resistance, axis=1)
res_min = np.nanmin(resistance, axis=1)

res_range = res_max - res_min
print("Jaguar Range: " + str((res_range[0]).round(2)))
print("Panthers Range: " + str((res_range[1]).round(2)))
```

```python
#IQR

res_Q1 = np.nanpercentile(resistance, 25, axis=1)
print("Jaguar Q1: " + str((res_Q1[0]).round(2)))
print("Panthers Q1: " + str((res_Q1[1]).round(2)))
```

```python
res_Q3 = np.nanpercentile(resistance,75, axis=1)
print("Jaguar Q3: " + str((res_Q3[0]).round(2)))
print("Panthers Q3: " + str((res_Q3[1]).round(2)))
```

```python
print(f'Jaguar IQR: ' + str((res_Q3[0] - res_Q1[0]).round(2)))
print(f'Panthers IQR: ' + str((res_Q3[1] - res_Q1[1]).round(2)))
```

```python
res_var = np.nanvar(resistance,ddof = 1, axis=1)
print("Jaguar Variance: " + str((res_var[0]).round(2)))
print("Panthers Variance: " + str((res_var[1]).round(2)))
```

```python
res_std = np.nanstd(resistance,ddof=1, axis=1)
print("Jaguar Standard Deviation: " + str((res_std[0]).round(2)))
print("Panthers Standard Deviation: " + str((res_std[1]).round( 2)))
```
