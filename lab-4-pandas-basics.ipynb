# Descriptive Statistics

```python
#Jaguar Vs Panthers
resistance = np.array([
    [10.6, 9.1, 9.3, 9.8, 10.5, 10.4, 9.5,11,10.4, 3, 9.8],
    [10.1, 11, 9.1, 20, 9.2, 10.8, 9.9, 9.2 , 9.1 ,9.1, np.nan]
]).T

resistance
```

```python
df = pd.DataFrame(resistance, columns=['Jaguar', 'Panthers'])
df
```

```python
type(df)
```

```python
df.info()
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.dropna()
```

```python
df.fillna(0)
```

```python
df['Panthers'].fillna(df['Panthers'].mean())
```

```python
df
```

```python
df.describe()
```

```python
jag_median = df['Jaguar'].median()
pan_median = df['Panthers'].median()
print(f'pan_median: {pan_median:.2f} "ohms"')
print(f'jag_median: {jag_median:.2f} "ohms"')
```

```python
jag_mode = df['Jaguar'].mode()
pan_mode = df['Panthers'].mode()

print("Jaguar Mode: ")

for i in jag_mode:
    print (i)

print("Panthers Mode: ")

for i in pan_mode:
    print (i)    
```

```python
Q1 = df['Jaguar'].quantile(0.25)
Q3 = df['Jaguar'].quantile(0.75)

print(f'Jaguar IQR = {Q3-Q1:.2f} "ohms"')
```

```python
Q1 = df['Panthers'].quantile(0.25)
Q3 = df['Panthers'].quantile(0.75)

print(f'Panthers IQR = {Q3-Q1:.2f} "ohms"')
```

# Bar Chart 

```python
jag_mean = df['Jaguar'].mean()
pan_mean = df['Panthers'].mean()

plt.figure()
bars = plt.bar(["Jaguar", "Panthers"],[jag_mean,pan_mean],
        color = 'red')
   
plt.title("Average Resistance By Machine")
plt.ylabel("Resistance (Ohms)")

plt.bar_label(bars)
plt.show()
```
